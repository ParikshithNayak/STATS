```python
#1 
import csv
import re
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, t


# ============================================================
# 0. CONFIGURATION
# ============================================================

N_VALUES = [20, 50, 100, 200, 1000]
K = 1000
CONFIDENCE_LEVEL = 0.95
ALPHA = 1.0 - CONFIDENCE_LEVEL
Z_CRITICAL = norm.ppf(1.0 - ALPHA / 2.0)

OUTPUT_DIR = Path("simulation_results")
OUTPUT_DIR.mkdir(exist_ok=True)

SHOW_PLOTS = True
DECIMALS = 4

rng = np.random.default_rng(42)


# ============================================================
# 1. FORMATTING HELPERS
# ============================================================

def fmt(value):
    """Format a numerical value to four decimal places."""
    return f"{value:.{DECIMALS}f}"


def safe_filename(text):
    """Create a filename-safe label."""
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", text).strip("_")


# ============================================================
# 2. GENERATE K INDEPENDENT SAMPLES OF SIZE n
# ============================================================

def generate_samples(distribution, n, **params):

    if distribution == "Gaussian":
        return rng.normal(
            loc=params["mu"],
            scale=params["sigma"],
            size=(K, n)
        )

    if distribution == "Uniform":
        return rng.uniform(
            low=params["low"],
            high=params["high"],
            size=(K, n)
        )

    if distribution == "Bernoulli":
        return rng.binomial(
            n=1,
            p=params["p"],
            size=(K, n)
        )

    raise ValueError(f"Unsupported distribution: {distribution}")


# ============================================================
# 3. SIMULATE ONE CONFIGURATION
# ============================================================

def run_experiment(
    distribution_name,
    distribution_label,
    true_mu,
    true_sigma,
    n,
    sigma_known,
    params
):
    """
    Run K experiments for one distribution and sample size.

    Produce:
      1. Histogram of sample means.
      2. Scatter plot of lower and upper confidence limits.
      3. Numerical calculations supporting the inference.
    """

    # --------------------------------------------------------
    # Step 1: Generate the observations
    # --------------------------------------------------------

    samples = generate_samples(distribution_name, n, **params)

    # Each row represents one experiment.
    sample_means = np.mean(samples, axis=1)

    # --------------------------------------------------------
    # Step 2: Calculate the sample standard deviations
    # --------------------------------------------------------

    sample_stds = np.std(samples, axis=1, ddof=1)

    # --------------------------------------------------------
    # Step 3: Calculate the standard error and critical value
    # --------------------------------------------------------

    theoretical_se = true_sigma / np.sqrt(n)

    if sigma_known:
        # Population sigma is known.
        critical_value = Z_CRITICAL
        standard_errors = np.full(K, theoretical_se)

        method = "Known population sigma (normal z interval)"

        # Theoretical width is fixed for a given n and sigma.
        theoretical_width = 2.0 * critical_value * theoretical_se

    else:
        # Population sigma is unknown.
        # Estimate it separately for every sample.
        critical_value = t.ppf(
            1.0 - ALPHA / 2.0,
            df=n - 1
        )

        standard_errors = sample_stds / np.sqrt(n)

        method = "Unknown population sigma (Student t interval)"

        # The interval widths now vary across experiments.
        theoretical_width = np.nan

    # --------------------------------------------------------
    # Step 4: Calculate the confidence interval limits
    # --------------------------------------------------------

    margins_of_error = critical_value * standard_errors

    lower = sample_means - margins_of_error
    upper = sample_means + margins_of_error

    interval_widths = upper - lower

    # --------------------------------------------------------
    # Step 5: Calculate empirical coverage
    # --------------------------------------------------------

    contains_mean = (lower <= true_mu) & (true_mu <= upper)

    coverage_count = int(np.sum(contains_mean))
    coverage_percentage = 100.0 * coverage_count / K

    expected_coverage_count = K * CONFIDENCE_LEVEL

    coverage_difference = (
        coverage_percentage - 100.0 * CONFIDENCE_LEVEL
    )

    # --------------------------------------------------------
    # Step 6: Supporting statistics
    # --------------------------------------------------------

    empirical_mean = np.mean(sample_means)
    empirical_sd = np.std(sample_means, ddof=1)

    average_se_used = np.mean(standard_errors)
    average_width = np.mean(interval_widths)

    # Theoretical SD of the sample mean:
    # sigma / sqrt(n)
    #
    # This is also the expected standard error for the
    # known-sigma method. With unknown sigma, the average
    # estimated SE need not equal it exactly.

    # --------------------------------------------------------
    # Step 7: Construct descriptive labels
    # --------------------------------------------------------

    sigma_status = (
        "sigma known"
        if sigma_known
        else "sigma estimated from each sample"
    )

    full_label = (
        f"{distribution_label} | {sigma_status} | n={n}"
    )

    # --------------------------------------------------------
    # Step 8: Plot histogram and confidence limits
    # --------------------------------------------------------

    fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

    # Histogram of the K sample means
    axes[0].hist(
        sample_means,
        bins=30,
        edgecolor="black",
        alpha=0.75
    )

    axes[0].axvline(
        true_mu,
        color="green",
        linestyle="--",
        linewidth=2,
        label=f"True mean = {fmt(true_mu)}"
    )

    axes[0].axvline(
        empirical_mean,
        color="black",
        linestyle=":",
        linewidth=1.5,
        label=f"Empirical mean = {fmt(empirical_mean)}"
    )

    axes[0].set_title(
        f"Sampling Distribution of the Mean\n"
        f"{full_label}"
    )
    axes[0].set_xlabel("Sample mean")
    axes[0].set_ylabel("Frequency")
    axes[0].legend(fontsize=8)
    axes[0].grid(alpha=0.2)

    # Scatter plot of the confidence limits
    experiment_numbers = np.arange(1, K + 1)

    axes[1].scatter(
        experiment_numbers,
        lower,
        color="blue",
        s=9,
        label="Lower confidence limit"
    )

    axes[1].scatter(
        experiment_numbers,
        upper,
        color="red",
        s=9,
        label="Upper confidence limit"
    )

    axes[1].axhline(
        true_mu,
        color="green",
        linestyle="--",
        linewidth=1.5,
        label=f"True mean = {fmt(true_mu)}"
    )

    axes[1].set_title(
        f"95% Confidence Interval Limits\n"
        f"{full_label}"
    )
    axes[1].set_xlabel("Experiment number")
    axes[1].set_ylabel("Confidence limit")
    axes[1].legend(fontsize=8)
    axes[1].grid(alpha=0.2)

    fig.suptitle(
        f"{distribution_label}\n"
        f"True sigma = {fmt(true_sigma)} | "
        f"Method: {method}",
        fontsize=11
    )

    fig.tight_layout(rect=[0, 0, 1, 0.88])

    filename = safe_filename(
        f"{distribution_label}_{sigma_status}_n{n}"
    )

    fig.savefig(
        OUTPUT_DIR / f"{filename}.png",
        dpi=150,
        bbox_inches="tight"
    )

    if SHOW_PLOTS:
        plt.show()

    plt.close(fig)

    # --------------------------------------------------------
    # Step 9: Print supporting calculations
    # --------------------------------------------------------

    print("\n" + "=" * 82)
    print(f"DISTRIBUTION: {distribution_label}")
    print(f"METHOD: {method}")
    print("=" * 82)

    print(f"Number of experiments (K)       : {K}")
    print(f"Observations per experiment (n) : {n}")
    print(f"True population mean (mu)       : {fmt(true_mu)}")
    print(f"True population sigma           : {fmt(true_sigma)}")
    print(f"Sigma used in interval          : "
          f"{'Known sigma' if sigma_known else 'Estimated sample sigma'}")

    print("\nSUPPORTING CALCULATIONS")
    print("-" * 82)

    print(
        f"Theoretical standard error, sigma/sqrt(n) : "
        f"{fmt(theoretical_se)}"
    )

    print(
        f"Empirical SD of sample means               : "
        f"{fmt(empirical_sd)}"
    )

    print(
        f"Critical value                              : "
        f"{fmt(critical_value)}"
    )

    print(
        f"Average standard error used                : "
        f"{fmt(average_se_used)}"
    )

    if sigma_known:
        print(
            f"Theoretical CI width                       : "
            f"{fmt(theoretical_width)}"
        )
    else:
        print(
            f"Theoretical fixed CI width                 : "
            f"Not fixed; depends on each sample's s"
        )

    print(
        f"Average observed CI width                  : "
        f"{fmt(average_width)}"
    )

    print(
        f"Empirical mean of sample means             : "
        f"{fmt(empirical_mean)}"
    )

    print("\nCOVERAGE RESULTS")
    print("-" * 82)

    print(
        f"Expected count at 95% coverage             : "
        f"{expected_coverage_count:.0f} out of {K}"
    )

    print(
        f"Observed intervals containing true mean    : "
        f"{coverage_count} out of {K}"
    )

    print(
        f"Observed coverage percentage               : "
        f"{coverage_percentage:.2f}%"
    )

    print(
        f"Difference from nominal 95%                : "
        f"{coverage_difference:+.2f} percentage points"
    )

    # --------------------------------------------------------
    # Step 10: Return results for the summary CSV
    # --------------------------------------------------------

    return {
        "distribution": distribution_label,
        "method": method,
        "n": n,
        "K": K,
        "true_mean": true_mu,
        "true_sigma": true_sigma,
        "theoretical_se": theoretical_se,
        "empirical_sd_of_means": empirical_sd,
        "critical_value": critical_value,
        "average_se_used": average_se_used,
        "theoretical_ci_width": theoretical_width,
        "average_ci_width": average_width,
        "empirical_mean_of_means": empirical_mean,
        "expected_coverage_count": int(expected_coverage_count),
        "observed_coverage_count": coverage_count,
        "coverage_percentage": coverage_percentage,
        "coverage_difference_pp": coverage_difference
    }


# ============================================================
# 4. RUN A COLLECTION OF EXPERIMENTS
# ============================================================

all_results = []


def run_collection(
    distribution_name,
    distribution_label,
    true_mu,
    true_sigma,
    sigma_known,
    params,
    sample_sizes=N_VALUES
):
    results = []

    for n in sample_sizes:
        result = run_experiment(
            distribution_name=distribution_name,
            distribution_label=distribution_label,
            true_mu=true_mu,
            true_sigma=true_sigma,
            n=n,
            sigma_known=sigma_known,
            params=params
        )

        results.append(result)
        all_results.append(result)

    return results


# ============================================================
# 5. GAUSSIAN DISTRIBUTION: KNOWN SIGMA
# ============================================================

print("\n\n")
print("#" * 82)
print("GAUSSIAN DISTRIBUTION — KNOWN POPULATION STANDARD DEVIATION")
print("#" * 82)

for sigma in [2, 3, 5]:
    run_collection(
        distribution_name="Gaussian",
        distribution_label=f"Gaussian N(0, {sigma}^2)",
        true_mu=0.0,
        true_sigma=float(sigma),
        sigma_known=True,
        params={"mu": 0.0, "sigma": float(sigma)}
    )


# ============================================================
# 6. GAUSSIAN DISTRIBUTION: UNKNOWN SIGMA
# ============================================================

print("\n\n")
print("#" * 82)
print("GAUSSIAN DISTRIBUTION — UNKNOWN POPULATION STANDARD DEVIATION")
print("#" * 82)

for sigma in [2, 3, 5]:
    run_collection(
        distribution_name="Gaussian",
        distribution_label=f"Gaussian N(0, {sigma}^2)",
        true_mu=0.0,
        true_sigma=float(sigma),
        sigma_known=False,
        params={"mu": 0.0, "sigma": float(sigma)}
    )


# ============================================================
# 7. UNIFORM DISTRIBUTIONS: KNOWN SIGMA
# ============================================================

print("\n\n")
print("#" * 82)
print("UNIFORM DISTRIBUTIONS — KNOWN POPULATION STANDARD DEVIATION")
print("#" * 82)

for low, high in [(0, 1), (-1, 1), (-2, 2)]:

    true_mu = (low + high) / 2.0
    true_sigma = (high - low) / np.sqrt(12.0)

    run_collection(
        distribution_name="Uniform",
        distribution_label=f"Uniform U({low}, {high})",
        true_mu=true_mu,
        true_sigma=true_sigma,
        sigma_known=True,
        params={"low": float(low), "high": float(high)}
    )


# ============================================================
# 8. BERNOULLI DISTRIBUTION: KNOWN SIGMA
# ============================================================

print("\n\n")
print("#" * 82)
print("BERNOULLI DISTRIBUTION — KNOWN POPULATION STANDARD DEVIATION")
print("#" * 82)

# Assumption: p = 0.5 because the question does not specify p.
p = 0.5

true_mu = p
true_sigma = np.sqrt(p * (1.0 - p))

run_collection(
    distribution_name="Bernoulli",
    distribution_label=f"Bernoulli(p={p:.3f})",
    true_mu=true_mu,
    true_sigma=true_sigma,
    sigma_known=True,
    params={"p": p}
)


# ============================================================
# 9. SAVE A CONSOLIDATED SUMMARY CSV
# ============================================================

summary_file = OUTPUT_DIR / "simulation_summary.csv"

with open(summary_file, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=list(all_results[0].keys())
    )

    writer.writeheader()

    for result in all_results:
        # Store numerical results at full precision in the CSV.
        writer.writerow(result)


# ============================================================
# 10. PRINT A COMPACT COMPARISON TABLE
# ============================================================

print("\n\n")
print("=" * 118)
print("FINAL COMPARISON: COVERAGE AND PRECISION")
print("=" * 118)

header = (
    f"{'Distribution':<28}"
    f"{'n':>6}"
    f"{'SE (theory)':>14}"
    f"{'SD (observed)':>16}"
    f"{'Avg CI width':>15}"
    f"{'Covered':>10}"
    f"{'Coverage %':>13}"
)

print(header)
print("-" * 118)

for r in all_results:
    print(
        f"{r['distribution']:<28}"
        f"{r['n']:>6d}"
        f"{r['theoretical_se']:>14.4f}"
        f"{r['empirical_sd_of_means']:>16.4f}"
        f"{r['average_ci_width']:>15.4f}"
        f"{r['observed_coverage_count']:>10d}"
        f"{r['coverage_percentage']:>12.2f}%"
    )

print("-" * 118)
print(f"Total configurations : {len(all_results)}")
print(f"Total intervals      : {len(all_results) * K}")
print(f"Plots folder         : {OUTPUT_DIR.resolve()}")
print(f"Summary CSV          : {summary_file.resolve()}")
```

```python
#2
import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from statistics import NormalDist

OUT_DIR = "resistor_quality_simulation"
os.makedirs(OUT_DIR, exist_ok=True)
mu, tolerance, allowed_outside = 100.0, 0.05, 0.01
lower, upper = mu*(1-tolerance), mu*(1+tolerance)
z = NormalDist().inv_cdf(1-allowed_outside/2)
sigma = (upper-mu)/z
N, B, M = 1_000_000, 1000, 1000
rng = np.random.default_rng(42)
x = rng.normal(mu, sigma, N)
outside = (x < lower) | (x > upper)
empirical = outside.mean()
theoretical = 2*(1-NormalDist().cdf((upper-mu)/sigma))

plt.figure(figsize=(9,5.5))
plt.hist(x[:200_000], bins=100, density=True, label="Simulated values")
xx = np.linspace(mu-4*sigma, mu+4*sigma, 500)
pdf = np.exp(-0.5*((xx-mu)/sigma)**2)/(sigma*np.sqrt(2*np.pi))
plt.plot(xx, pdf, linewidth=2, label="Theoretical Gaussian density")
plt.axvline(lower, linestyle="--", label=f"Lower limit: {lower:.0f} Ω")
plt.axvline(upper, linestyle="--", label=f"Upper limit: {upper:.0f} Ω")
plt.axvline(mu, linestyle=":", label=f"Mean: {mu:.0f} Ω")
plt.title(f"Resistance Distribution: μ={mu:.0f} Ω, σ={sigma:.4f} Ω")
plt.xlabel("Resistance (Ω)"); plt.ylabel("Density"); plt.legend(); plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"resistance_distribution.png"),dpi=150); plt.close()

batches = rng.normal(mu, sigma, size=(B,M))
batch_pct = (((batches < lower) | (batches > upper)).mean(axis=1))*100
plt.figure(figsize=(9,5.5))
plt.hist(batch_pct,bins=30)
plt.axvline(1,linestyle="--",linewidth=2,label="Allowed maximum: 1%")
plt.axvline(empirical*100,linestyle=":",linewidth=2,label=f"Overall: {empirical*100:.3f}%")
plt.title("Outside-Tolerance Percentage Across 1,000 Batches")
plt.xlabel("Outside-tolerance resistors per batch (%)"); plt.ylabel("Number of batches")
plt.legend(); plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"batch_defect_percentages.png"),dpi=150); plt.close()

running = np.cumsum(outside)/np.arange(1,N+1)*100
points = np.unique(np.logspace(2,6,140).astype(int))
plt.figure(figsize=(9,5.5))
plt.plot(points,running[points-1],linewidth=1.8)
plt.axhline(1,linestyle="--",linewidth=2,label="Target: 1%")
plt.xscale("log"); plt.title("Running Empirical Percentage Outside Tolerance")
plt.xlabel("Number simulated (log scale)"); plt.ylabel("Outside tolerance (%)")
plt.legend(); plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"running_defect_proportion.png"),dpi=150); plt.close()

summary = pd.DataFrame([
    ("Maximum sigma (Ω)",sigma),("Theoretical outside (%)",theoretical*100),
    ("Simulated resistors",N),("Outside count",int(outside.sum())),
    ("Empirical outside (%)",empirical*100),("Empirical within (%)",(1-empirical)*100),
    ("Mean batch outside (%)",batch_pct.mean())
],columns=["Metric","Value"])
summary.to_csv(os.path.join(OUT_DIR,"simulation_summary.csv"),index=False)
print(f"Tolerance band: [{lower:.1f}, {upper:.1f}] Ω")
print(f"Maximum sigma: {sigma:.6f} Ω")
print(f"Theoretical outside tolerance: {theoretical*100:.4f}%")
print(f"Simulated outside tolerance: {int(outside.sum())}/{N} = {empirical*100:.4f}%")
print(f"Simulated within tolerance: {(1-empirical)*100:.4f}%")
print(f"Mean batch outside percentage: {batch_pct.mean():.4f}%")
print(f"Files saved in {OUT_DIR}")
```
