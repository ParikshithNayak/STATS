# Bivariate Gaussian Random Vector: Two-Sensor Example

Suppose two temperature sensors are installed close to each other in the same industrial process.

Let

\[
X=\text{error in Sensor 1}, \qquad
Y=\text{error in Sensor 2}.
\]

We study the two errors together as the random vector

\[
\mathbf{Z}=
\begin{bmatrix}
X\\
Y
\end{bmatrix}.
\]

The reason for studying them jointly is that both sensors may be affected by common operating or environmental conditions, so their errors may be correlated.

## Mean vector and covariance matrix

For a simple standard example, assume that both sensors are unbiased:

\[
E[X]=E[Y]=0.
\]

Also assume

\[
\operatorname{Var}(X)=\operatorname{Var}(Y)=1.
\]

Therefore,

\[
\boldsymbol{\mu}=
\begin{bmatrix}
0\\
0
\end{bmatrix}.
\]

Let

\[
\rho=\operatorname{Corr}(X,Y), \qquad -1<\rho<1.
\]

Since both standard deviations are 1,

\[
\operatorname{Cov}(X,Y)=\rho.
\]

Hence,

\[
\boldsymbol{\Sigma}=
\begin{bmatrix}
1 & \rho\\
\rho & 1
\end{bmatrix}.
\]

## Bivariate Gaussian model

Assume that the sensor-error vector is jointly Gaussian:

\[
\begin{bmatrix}
X\\
Y
\end{bmatrix}
\sim
\mathcal{N}
\left(
\begin{bmatrix}
0\\
0
\end{bmatrix},
\begin{bmatrix}
1 & \rho\\
\rho & 1
\end{bmatrix}
\right).
\]

The joint PDF is

\[
f_{XY}(x,y)=
\frac{1}{2\pi\sqrt{1-\rho^2}}
\exp\left[
-\frac{x^2-2\rho xy+y^2}
{2(1-\rho^2)}
\right].
\]

### Meaning of \(\rho\)

- \(\rho>0\): the two sensor errors tend to move in the same direction.
- \(\rho=0\): no linear correlation. For jointly Gaussian variables, this also means independence.
- \(\rho<0\): the two sensor errors tend to move in opposite directions.

Now use the slider below and observe how the **joint PDF changes as \(\rho\) changes**.

```python
# Run this cell once before using the slider

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Grid for the joint PDF
x = np.linspace(-4, 4, 120)
y = np.linspace(-4, 4, 120)
X_grid, Y_grid = np.meshgrid(x, y)

def bivariate_gaussian_pdf(x, y, rho):
    coefficient = 1 / (2 * np.pi * np.sqrt(1 - rho**2))
    exponent = -(x**2 - 2*rho*x*y + y**2) / (2 * (1 - rho**2))
    return coefficient * np.exp(exponent)

print("Setup complete. Run the next cell for the interactive slider.")
```

## Interactive demonstration: change \(\rho\)

Move the slider from negative to positive values.

The left figure shows the **3D joint PDF** and the right figure shows its **contours**.

Pay attention to the direction and shape of the contours.

```python
rho_slider = widgets.FloatSlider(
    value=0.0,
    min=-0.95,
    max=0.95,
    step=0.05,
    description=r'\(\rho\)',
    continuous_update=True,
    readout_format='.2f',
    style={'description_width': '40px'},
    layout=widgets.Layout(width='70%')
)

plot_output = widgets.Output()

def update_pdf(change=None):
    rho = rho_slider.value
    Z = bivariate_gaussian_pdf(X_grid, Y_grid, rho)

    with plot_output:
        clear_output(wait=True)

        # 3D PDF
        fig = plt.figure(figsize=(13, 5))

        ax1 = fig.add_subplot(1, 2, 1, projection='3d')
        ax1.plot_surface(X_grid, Y_grid, Z, cmap='viridis', edgecolor='none')
        ax1.set_xlabel('Sensor 1 error, x')
        ax1.set_ylabel('Sensor 2 error, y')
        ax1.set_zlabel(r'$f_{XY}(x,y)$')
        ax1.set_title(fr'Joint PDF: $\rho={rho:.2f}$')
        ax1.set_zlim(0, 0.55)

        # Contour plot
        ax2 = fig.add_subplot(1, 2, 2)
        contour = ax2.contour(X_grid, Y_grid, Z, levels=12, cmap='viridis')
        ax2.clabel(contour, inline=True, fontsize=8)
        ax2.set_xlabel('Sensor 1 error, x')
        ax2.set_ylabel('Sensor 2 error, y')
        ax2.set_title(fr'PDF Contours: $\rho={rho:.2f}$')
        ax2.set_xlim(-4, 4)
        ax2.set_ylim(-4, 4)
        ax2.set_aspect('equal', adjustable='box')

        plt.tight_layout()
        plt.show()

rho_slider.observe(update_pdf, names='value')

display(rho_slider, plot_output)
update_pdf()
```

## What should students observe?

### When \(\rho=0\)

The contours are circular because

\[
\operatorname{Var}(X)=\operatorname{Var}(Y)=1
\]

and there is no preferred direction.

### When \(\rho>0\)

The contours become ellipses tilted upward.

A positive error in Sensor 1 tends to occur together with a positive error in Sensor 2.

### When \(\rho<0\)

The contours tilt downward.

A positive error in one sensor tends to occur with a negative error in the other.

### As \(|\rho|\) approaches 1

The ellipse becomes narrower.

This indicates a stronger linear relationship between the two sensor errors.

## Connection with independence

For \(\rho=0\),

\[
f_{XY}(x,y)
=
\frac{1}{2\pi}
\exp\left[-\frac{x^2+y^2}{2}\right].
\]

This can be written as

\[
f_{XY}(x,y)
=
\left[
\frac{1}{\sqrt{2\pi}}e^{-x^2/2}
\right]
\left[
\frac{1}{\sqrt{2\pi}}e^{-y^2/2}
\right].
\]

Hence,

\[
f_{XY}(x,y)=f_X(x)f_Y(y).
\]

Therefore, for a **jointly Gaussian** pair,

\[
\rho=0
\quad\Longrightarrow\quad
X \text{ and } Y \text{ are independent}.
\]
